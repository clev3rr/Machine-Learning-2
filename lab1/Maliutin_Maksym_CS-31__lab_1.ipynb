# Theory: How to use backpropagation on a single perceptron?

## I. Equation for `binary cross-entropy`:

$$
Loss = - \frac{1}{N} \sum_{i=1}^{N} [y_i * log(y^*_i) + (1-y_i) * log(1 - y^*_i)]
$$
, where:
 
* $ N $ - number of samples
* $ y_i $ - probability of correct class (in case of binary problem 0 for 0-th class and 1 for 1-st class)
* $ y^*_i $ - collected value of class probability (by your neural network or any other kind of NN)

## II. Equation for a single perceptron:

For a single perceptron with two inputs ( $x_1$ and $x_2$), the forward pass consist of two steps: the linear sum and the activation function

1. Linear sum:

$$
z = x_1 * w_1 + x_2 * w_2 + b
$$

2. Sigmoid activation function:

$$
y_{pred} = \sigma(z) = \frac{1}{1 + exp(-z)}
$$


## III. Loss function for a single perceptron:

For a single example $X=(x_1, x_2)$, $y_{true}$ the `binary cross-entropy` will look like this: 

$$
Loss = -[y_{true} * log(y_{pred}) + (1 - y_{true}) * log(1 - y_{pred})]
$$

The idea of backpropagation is to use the chain rule of calculus to find the gradient, which tells us the influence of each weight on the final loss.

And for each weight adjustment temp will look like this:

$$
\frac{\delta L}{\delta w_1} = \frac{\delta L}{\delta y_{pred}} * \frac{\delta y_{pred}}{\delta z} * \frac{\delta z}{\delta w_1}
$$

where $z$ = $x_1 * w_1 + x_2 * w_2 + b$  -- linear sum.

Let's break down the partial derivatives:

1. Derivative of the loss function with respect to the prediction:

$$
\frac{\delta L}{\delta y_{pred}} = \frac{y_{pred} - y_{true}}{y_{pred}(1 - y_{pred})}
$$

2. Derivative of the sigmoid activation with respect to the linear sum:

$$
\frac{\delta y_{pred}}{\delta z} = y_{pred} * (1 - y_{pred})
$$

3. Derivative of the linear sum with respect to the weights:

$$
\frac{\delta z}{\delta w_1} = x_1
$$


Calculating the final gradients:

When we multiply these together, notice how perfectly the denominator of the loss derivative cancels out the sigmoid derivative! This is the mathematical advantage of pairing sigmoid with binary cross-entropy.

$$
\triangle w_1 = \frac{\delta L}{\delta w_1} = \frac{y_{pred} - y_{true}}{y_{pred}(1 - y_{pred})} * y_{pred} * (1 - y_{pred}) * x_1 = [y_{pred} - y_{true}] * x_1
$$

$$
\triangle w_2 = \frac{\delta L}{\delta w_2} = \frac{y_{pred} - y_{true}}{y_{pred}(1 - y_{pred})} * y_{pred} * (1 - y_{pred}) * x_2 = [y_{pred} - y_{true}] * x_2
$$

$$
\triangle b = \frac{\delta L}{\delta b} = \frac{y_{pred} - y_{true}}{y_{pred}(1 - y_{pred})} * y_{pred} * (1 - y_{pred}) * 1 = = [y_{pred} - y_{true}] * 1
$$


So adjusted weights to minimize loss value would be:

$$
w_1^{new} = w_1 - a * \triangle w_1
$$

$$
w_2^{new} = w_2 - a * \triangle w_2
$$

$$
b^{new} = b - a * \triangle b
$$


In case of multiple training samples $X = {(x_{1, 1}, x_{1, 2}), (x_{2, 1}, x_{2, 2}), ..., (x_{n, 1}, x_{n, 2})}; y = {y_1, y_2, ..., y_n}$ calculate the gradients for each sample and average them (since our original loss function in Section I divides by $N$) before applying the final weight update.

```python
def relu(x):
    return np.maximum(x, 0.0)


def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-np.clip(x, -50, 50)))


class NeuralNetwork:
    def __init__(self, input_size, hidden_sizes=(32, 16), output_size=1, seed=42):
        rng = np.random.default_rng(seed)
        h0, h1 = hidden_sizes

        self.W0 = rng.normal(0, 0.05, size=(input_size, h0)).astype(np.float32)
        self.b0 = np.zeros((1, h0), dtype=np.float32)

        self.W1 = rng.normal(0, 0.05, size=(h0, h1)).astype(np.float32)
        self.b1 = np.zeros((1, h1), dtype=np.float32)

        self.W2 = rng.normal(0, 0.05, size=(h1, output_size)).astype(np.float32)
        self.b2 = np.zeros((1, output_size), dtype=np.float32)

    def forward(self, X):
        self.Z0 = X @ self.W0 + self.b0
        self.A0 = relu(self.Z0)

        self.Z1 = self.A0 @ self.W1 + self.b1
        self.A1 = relu(self.Z1)

        self.Z2 = self.A1 @ self.W2 + self.b2
        self.A2 = sigmoid(self.Z2)
        return self.A2

    def backward(self, X, y, learning_rate=0.01):
        n = X.shape[0]

        dZ2 = self.A2 - y
        dW2 = (self.A1.T @ dZ2) / n
        db2 = np.sum(dZ2, axis=0, keepdims=True) / n

        dA1 = dZ2 @ self.W2.T
        dZ1 = dA1 * (self.Z1 > 0)
        dW1 = (self.A0.T @ dZ1) / n
        db1 = np.sum(dZ1, axis=0, keepdims=True) / n

        dA0 = dZ1 @ self.W1.T
        dZ0 = dA0 * (self.Z0 > 0)
        dW0 = (X.T @ dZ0) / n
        db0 = np.sum(dZ0, axis=0, keepdims=True) / n

        self.W2 -= learning_rate * dW2
        self.b2 -= learning_rate * db2
        self.W1 -= learning_rate * dW1
        self.b1 -= learning_rate * db1
        self.W0 -= learning_rate * dW0
        self.b0 -= learning_rate * db0
```

```python
def compute_binary_cross_entropy_loss(y_true, y_pred):
    eps = 1e-8
    y_pred = np.clip(y_pred, eps, 1.0 - eps)
    return float(-np.mean(y_true * np.log(y_pred) + (1.0 - y_true) * np.log(1.0 - y_pred)))
```

```python
def compute_accuracy(y_true, y_pred_proba, threshold=0.5):
    y_hat = (y_pred_proba >= threshold).astype(np.float32)
    return float(np.mean(y_hat == y_true))


def compute_precision_recall_f1(y_true, y_pred_proba, threshold=0.5):
    y_hat = (y_pred_proba >= threshold).astype(np.float32)

    tp = np.sum((y_hat == 1) & (y_true == 1))
    fp = np.sum((y_hat == 1) & (y_true == 0))
    fn = np.sum((y_hat == 0) & (y_true == 1))

    precision = tp / max(tp + fp, 1)
    recall = tp / max(tp + fn, 1)
    f1 = 2 * precision * recall / max(precision + recall, 1e-8)
    return float(precision), float(recall), float(f1)
```

```python
def train_numpy_model(
    hidden_sizes=(32, 16),
    learning_rate=0.01,
    num_epochs=600,
    verbose_step=100,
    ):
    model = NeuralNetwork(
        input_size=input_size,
        hidden_sizes=hidden_sizes,
        output_size=1,
        seed=RANDOM_STATE,
        )

    train_loss_history = []
    train_acc_history = []
    val_loss_history = []
    val_acc_history = []

    for epoch in tqdm(range(1, num_epochs + 1)):
        y_pred_train = model.forward(x_train)
        train_loss = compute_binary_cross_entropy_loss(y_train, y_pred_train)
        train_acc = compute_accuracy(y_train, y_pred_train)

        model.backward(x_train, y_train, learning_rate=learning_rate)

        y_pred_val = model.forward(x_test)
        val_loss = compute_binary_cross_entropy_loss(y_test, y_pred_val)
        val_acc = compute_accuracy(y_test, y_pred_val)

        train_loss_history.append(train_loss)
        train_acc_history.append(train_acc)
        val_loss_history.append(val_loss)
        val_acc_history.append(val_acc)

        if verbose_step and epoch % verbose_step == 0:
            print(
                f"Epoch {epoch:4d} | "
                f"train_loss={train_loss:.4f}, train_acc={train_acc:.4f}, "
                f"val_loss={val_loss:.4f}, val_acc={val_acc:.4f}"
            )

    y_pred_test = model.forward(x_test)
    precision, recall, f1 = compute_precision_recall_f1(y_test, y_pred_test)

    history = {
        "train_loss": train_loss_history,
        "train_acc": train_acc_history,
        "val_loss": val_loss_history,
        "val_acc": val_acc_history,
    }
    metrics = {
        "test_loss": compute_binary_cross_entropy_loss(y_test, y_pred_test),
        "test_acc": compute_accuracy(y_test, y_pred_test),
        "precision": precision,
        "recall": recall,
        "f1": f1,
    }
    return model, history, metrics
```

# II. Using library for implementation your models (Tensorflow & Keras)

```python
initializer = tf.keras.initializers.RandomNormal(mean=0.0, stddev=0.05)
```

```python
def build_tf_model(input_dim, hidden_sizes=(32, 16), dropout=0.0):
    model = tf.keras.Sequential()
    model.add(tf.keras.layers.Input(shape=(input_dim,)))

    for i, units in enumerate(hidden_sizes):
        model.add(
            tf.keras.layers.Dense(
                units=units,
                activation="relu",
                kernel_initializer=initializer,
                bias_initializer="zeros",
                name=f"Hidden_{i}",
            )
        )
        if dropout > 0:
            model.add(tf.keras.layers.Dropout(dropout))

    model.add(
        tf.keras.layers.Dense(
            units=1,
            activation="sigmoid",
            kernel_initializer=initializer,
            bias_initializer="zeros",
            name="Output",
        )
    )
    return model
```

```python
def train_tf_model(hidden_sizes=(32, 16), learning_rate=0.001, epochs=40, batch_size=128, dropout=0.0):
    model = build_tf_model(input_size, hidden_sizes=hidden_sizes, dropout=dropout)
    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)

    model.compile(
        optimizer=optimizer,
        loss=tf.keras.losses.BinaryCrossentropy(),
        metrics=[
            tf.keras.metrics.BinaryAccuracy(name="accuracy"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
        ],
    )

    early_stopping = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True,
    )

    history = model.fit(
        x_train,
        y_train,
        validation_data=(x_test, y_test),
        epochs=epochs,
        batch_size=batch_size,
        verbose=0,
        callbacks=[early_stopping],
        class_weight=class_weights,
    )

    eval_result = model.evaluate(x_test, y_test, verbose=0, return_dict=True)
    metrics = {
        "loss": float(eval_result.get("loss", np.nan)),
        "accuracy": float(eval_result.get("accuracy", eval_result.get("binary_accuracy", np.nan))),
        "precision": float(eval_result.get("precision", np.nan)),
        "recall": float(eval_result.get("recall", np.nan)),
    }

    pred_test = model.predict(x_test, verbose=0)
    precision, recall, f1 = compute_precision_recall_f1(y_test, pred_test)
    metrics["precision"] = precision if np.isnan(metrics["precision"]) else metrics["precision"]
    metrics["recall"] = recall if np.isnan(metrics["recall"]) else metrics["recall"]
    metrics["f1"] = f1
    return model, history, metrics
```
